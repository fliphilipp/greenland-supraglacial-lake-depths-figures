## for generating contour lines from ArcticDEM

```python
import rasterio
from rasterio.windows import from_bounds
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
from shapely.geometry import LineString, mapping, MultiLineString
import fiona
from fiona.crs import from_epsg

# Define the coordinates for the window in the DEM's CRS
xlim_gris_dem = [-688000.0, 890000.0]
ylim_gris_dem = [-3400000.0, -582000.0]

# Open the DEM file
with rasterio.open('data/arcticDEM/arcticdem_mosaic_1km_v4.1_dem.tif') as src:

    crs_dem = src.crs
    
    # Step 1: Define the window based on the bounding box in CRS coordinates
    window = from_bounds(
        left=xlim_gris_dem[0],
        bottom=ylim_gris_dem[0],
        right=xlim_gris_dem[1],
        top=ylim_gris_dem[1],
        transform=src.transform
    )
    
    # Step 2: Read the data within this window
    data = src.read(1, window=window)
    
    # Step 3: Get the transform for the window
    window_transform = src.window_transform(window)
    
    # Get the coordinates
    x = np.arange(data.shape[1])
    y = np.arange(data.shape[0])
    x, y = np.meshgrid(x, y)
    x, y = rasterio.transform.xy(window_transform, y, x, offset='center')

# Convert coordinates to numpy arrays
x = np.array(x)
y = np.array(y)

# Step 4: Smooth the data (optional)
data_smooth = gaussian_filter(data, sigma=10)

# Step 5: Generate contour lines
fig, ax = plt.subplots()
contour_levels = [500, 1000, 1500, 2000, 2200, 2500, 3000, 3500, 4000] 
contours = ax.contour(x, y, data_smooth, levels=contour_levels)
plt.close(fig)

# Initialize a list to hold dictionaries with 'geometry' and 'level' keys
contour_data = []

# Iterate over each contour collection and group lines by level
for level, collection in zip(contour_levels, contours.collections):
    lines = []
    for path in collection.get_paths():
        v = path.vertices
        line = LineString(v)
        lines.append(line)
    
    # Create a MultiLineString for this contour level and add to the data list
    if lines:
        multiline = MultiLineString(lines)
        contour_data.append({'elevation': level, 'geometry': multiline})

# Convert the list of dictionaries to a GeoDataFrame
gdf_contours = gpd.GeoDataFrame(contour_data, crs=crs_dem)

fig, ax = plt.subplots(figsize=[6,9])
gre_coastline.plot(ax=ax, facecolor=[0.75]*3, edgecolor='none', lw=0.5, zorder=-3)
gre_coastline.boundary.plot(ax=ax, color='black', lw=0.5, zorder=0)
gdf_basins.plot(color='white', ax=ax, lw=0.6, zorder=-2, alpha=1)
gdf_basins.plot(color=ice_color, ax=ax, lw=0.6, zorder=-1, alpha=0.15)
gdf_basins.boundary.plot(color=gris_color, ax=ax, lw=0.6, zorder=10)
gdf_contours[gdf_contours.elevation != 2200].clip(gdf_basins).plot(ax=ax, column='elevation', cmap=cmc.hawaii, lw=0.5)
# gdf_contours[gdf_contours.elevation != 2200].clip(gdf_basins).plot(ax=ax, color='gray', lw=0.5)
display(fig)

gdf_contours.to_file('data/shapefiles/gris_contours.shp')
```
